## Q 1.3.7

```python
gift_crosstab=pd.crosstab(gifts['Country of Giftor'],gifts['Institution Name'],
            values=gifts['Foreign Gift Amount'],
            aggfunc=sum)
gift_crosstab
```

```python
import plotly.graph_objects as go

df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()

#giftor = 'Giftor Name'
giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 25

flows = (
    df.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
```

>The alluvial plot can also show us a lot of what we need. While, the plot itself does not give a concrete answer, it does give us a clear top choices to analyze. We can see that some of the largest flows are from Qatar to Cornell, Texas A&M, and Carnegie Mellon and Bermuda has a large flow to Carnegie Mellon. Now, we can find out which is the largest.

```python
gift_crosstab.loc['QATAR','Carnegie Mellon University']
#largest_gifts={'Qatar to CMU':gift_crosstab.loc['QATAR','Carnegie Mellon University'],
#              'Qatar to Texas A&M': gift_crosstab.loc['QATAR','Texas A&M University'],
#              'Qatar to Cornell':gift_crosstab.loc['QATAR','Cornell University'],
#              'Bermuda to CMU':gift_crosstab.loc['BERMUDA','Carnegie Mellon University']}
#largest_gifts=pd.DataFrame(largest_gifts)

pd.DataFrame([gift_crosstab.loc['QATAR','Carnegie Mellon University'],gift_crosstab.loc['QATAR','Texas A&M University'],
            gift_crosstab.loc['QATAR','Cornell University'],gift_crosstab.loc['BERMUDA','Carnegie Mellon University']],
            index=['Qatar to CMU','Qatar to Texas A&M','Qatar to Cornell','Bermuda to CMU']).rank(0, ascending=False)
```

>Now we get our concrete answer from the our prediction based on the graph. Qatar to Cornell is the largest flow of money followed by Bermuda to CMU, Qatar to Texas A&M, and then Qatar to CMU.
