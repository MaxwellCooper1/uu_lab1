```python
import numpy as np
import pandas as pd
import plotly.express as px

gifts=pd.read_csv('ForeignGifts_edu.csv')
```

```python
pd.set_option('display.max_rows',100)
gifts_to_uva=gifts[gifts['Institution Name']=="University of Virginia"]
```

```python
gifts_to_uva[['Country of Giftor','Foreign Gift Amount', 'Gift Type', 'Giftor Name']]
```

>Here, I'm making a new dataframe that is grouped by not only the country, but where else they donated to and how much. This is going to be perfect for data exploration.

```python
country_to_school=gifts.groupby(['Country of Giftor','Institution Name']).sum()['Foreign Gift Amount']
```

```python
country_to_school=pd.DataFrame(country_to_school)
country_to_school[1:50]

#since this is long, we look at it in chunks
```

```python
country_to_school.xs("AUSTRALIA",level="Country of Giftor")
```

>Wow, australia likes to give a lot of money to a lot of different schools. Specifically, they like to give to a lot of big research schools and they're pretty familiar with giving to virginia schools.

Next, seperate out for just australia and look to see who is giving. It's not helpful to talk to the country as a whole. Who specifically do we need to talk to?

```python
australian_giftors=gifts.loc[gifts['Country of Giftor']=='AUSTRALIA'].groupby(gifts['Giftor Name']).sum()[['Institution Name', 'Foreign Gift Amount']]
australian_giftors.head(20)
```

> After discovering some patterns in this data, I'm seeing a TON of money is going into health research (which is great since UVA has a huge health research department)
